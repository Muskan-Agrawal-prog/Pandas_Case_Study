### Part - 7 [CREATING NEW COLUMNS]

```python
#1. Create a new column called Annual_Salary from the monthly Salary.
df["Annual_Salary"] = df["Salary"]*12
```

```python
#2. Create a new column called Performance_Level based on Performance_Score: 90+ = Excellent, 80–89 = Good, below 80 = Needs Improvement.
df["Performance_Level"] = df["Performance_Score"].apply( lambda x : 'Exellent' if x >=90 else 'Good' if x >80  else 'Needs Improvement')
```

```python
#3. Create a new column called Experience_Level: 8+ years = Senior, 4–7 years = Mid-Level, below 4 years =Junior.
df["Experience_Level"] = df["Experience"].apply( lambda x: 'Senior' if x >=8 else 'mid-Level' if x >4 else 'Junior' )
```

```python
#4. Create a new column called Attendance_Status: 90+ = Good, below 90 = Low.
df["Attendance_Status"] = df["Attendance"].apply(lambda x: 'Good' if x >= 90 else 'low')
```
